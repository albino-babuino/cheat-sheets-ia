# PyCharm — Cheat Sheet detallada

Referencia rápida y exhaustiva del IDE **PyCharm** (JetBrains) para desarrollo en Python: atajos, navegación, edición, depuración, Git, Jupyter y configuración.

**Versión de referencia:** PyCharm 2024.x (Community o Professional). Atajos indicados para **Linux** (en Windows/Mac algunos usan Ctrl/Cmd o Alt/Option de forma equivalente).

---
## 1. Interfaz y ventanas principales

| Elemento | Descripción | Cómo abrirlo/cerrarlo |
|----------|-------------|------------------------|
| **Project** | Árbol del proyecto (archivos y carpetas) | Barra lateral izquierda; `Alt+1` |
| **Editor** | Archivo actual (código) | Centro; doble clic en archivo |
| **Terminal** | Consola integrada (bash, etc.) | `Alt+F12` o **View → Tool Windows → Terminal** |
| **Run** | Salida de ejecuciones y logs | `Alt+4` o barra inferior **Run** |
| **Debug** | Breakpoints, variables, consola de depuración | `Alt+5` (visible al depurar) |
| **Structure** | Estructura del archivo (clases, funciones) | `Alt+7` |
| **Problems / TODO** | Errores, advertencias, TODOs | Barra inferior **Problems** |
| **Find** | Búsqueda en proyecto | `Ctrl+Shift+F` |

**Consejo:** Cerrar ventanas con el icono de la esquina o `Shift+Esc` (deja solo el editor). Restaurar con el mismo atajo o desde **View → Tool Windows**.

## 2. Navegación y búsqueda

### Ir a archivo, clase, símbolo

| Atajo | Acción |
|-------|--------|
| **Doble `Shift`** | Buscar en todo (archivos, clases, acciones, símbolos) |
| **`Ctrl+N`** | Ir a clase (por nombre) |
| **`Ctrl+Shift+N`** | Ir a archivo (por nombre) |
| **`Ctrl+Alt+Shift+N`** | Ir a símbolo (función, variable, etc.) |
| **`Ctrl+B`** o **Ctrl+Click** | Ir a la definición del símbolo bajo el cursor |
| **`Ctrl+Alt+B`** | Ir a implementación (si es una interfaz/abstracto) |
| **`Alt+F7`** | Buscar usos del símbolo (Find Usages) |
| **`Ctrl+H`** (en búsqueda) | Jerarquía de herencia de la clase |

### Navegar en el historial del cursor

| Atajo | Acción |
|-------|--------|
| **`Ctrl+Alt+←`** | Volver atrás (posición anterior del cursor) |
| **`Ctrl+Alt+→`** | Ir adelante |
| **`Ctrl+E`** | Archivos recientes |
| **`Ctrl+Shift+E`** | Ubicaciones recientes (archivo + línea) |

### Buscar y reemplazar

| Atajo | Acción |
|-------|--------|
| **`Ctrl+F`** | Buscar en el archivo actual |
| **`Ctrl+R`** | Buscar y reemplazar en el archivo |
| **`Ctrl+Shift+F`** | Buscar en todo el proyecto (path) |
| **`Ctrl+Shift+R`** | Reemplazar en todo el proyecto |
| **`Ctrl+G`** | Ir a línea (y columna) |

**Buscar en proyecto:** se puede restringir por tipo de archivo (*.py), carpeta, y usar expresiones regulares.

## 3. Edición de código

### Comentarios y formato

| Atajo | Acción |
|-------|--------|
| **`Ctrl+/`** | Comentar/descomentar línea(s) con `#` |
| **`Ctrl+Shift+/`** | Comentar bloque (docstring/comentario de bloque) |
| **`Ctrl+Alt+L`** | Reformatear código (PEP 8 según configuración) |
| **`Ctrl+Alt+I`** | Auto-indentar líneas seleccionadas |

### Duplicar, mover, borrar

| Atajo | Acción |
|-------|--------|
| **`Ctrl+D`** | Duplicar línea o selección |
| **`Ctrl+Y`** | Eliminar línea (sin copiar al portapapeles) |
| **`Ctrl+Shift+↑/↓`** | Mover línea(s) arriba/abajo |
| **`Shift+Enter`** | Nueva línea abajo (y salir de la actual) |
| **`Ctrl+Shift+Enter`** | Completar línea (añadir `:` si hace falta, etc.) |

### Selección y múltiples cursores

| Atajo | Acción |
|-------|--------|
| **`Ctrl+W`** | Ir ampliando selección (palabra → expresión → línea → bloque) |
| **`Ctrl+Shift+W`** | Reducir selección |
| **`Alt+J`** | Seleccionar siguiente ocurrencia de la palabra (multi-cursor) |
| **`Alt+Shift+J`** | Seleccionar todas las ocurrencias |
| **`Ctrl+Ctrl`** (mantener) + ↑/↓ | Añadir cursor arriba/abajo |

### Plegar/desplegar código (folding)

| Atajo | Acción |
|-------|--------|
| **`Ctrl+Shift+.`** | Plegar bloque actual |
| **`Ctrl+Shift+,`** | Desplegar bloque actual |
| **`Ctrl+Shift+Alt+.`** | Plegar todo |
| **`Ctrl+Shift+Alt+,`** | Desplegar todo |

## 4. Completado e inspecciones

### Autocompletado (code completion)

| Atajo | Acción |
|-------|--------|
| **`Ctrl+Space`** | Completado básico (variables, nombres) |
| **`Ctrl+Shift+Space`** | Completado por tipo (sugiere según contexto) |
| **`Ctrl+P`** (dentro de paréntesis) | Información de parámetros de la función |
| **`Ctrl+Q`** | Documentación rápida (docstring/signature) |
| **`Ctrl+Shift+I`** | Ver definición en ventana flotante (quick definition) |

### Acciones rápidas (intention actions)

| Atajo | Acción |
|-------|--------|
| **`Alt+Enter`** | Mostrar acciones disponibles (importar, crear variable, corregir, etc.) |

Ejemplos: "Import this name", "Create function", "Add type hint", "Rename", "Extract variable/method", "Implement missing methods".

### Inspecciones y errores

- Los subrayados en el editor indican errores (rojo), advertencias (amarillo) y sugerencias (gris).
- **`Alt+Enter`** sobre el subrayado suele proponer la corrección.
- **Settings → Editor → Inspections** permite activar/desactivar inspecciones por categoría (Python, PEP 8, etc.).

## 5. Refactorización

| Atajo | Acción |
|-------|--------|
| **`Ctrl+Alt+Shift+T`** | Menú "Refactor This" (elegir tipo de refactor) |
| **`Shift+F6`** | Renombrar (símbolo, archivo, etc.) — actualiza todas las referencias |
| **`Ctrl+Alt+M`** | Extraer método (selección → nueva función) |
| **`Ctrl+Alt+V`** | Extraer variable (selección → nueva variable) |
| **`Ctrl+Alt+C`** | Extraer constante |
| **`Ctrl+Alt+P`** | Extraer parámetro (introducir argumento en función) |
| **`F5`** | Copiar archivo/clase/función |
| **`F6`** | Mover archivo/clase (cambiar ubicación y actualizar imports) |
| **`Alt+Delete`** | Eliminar de forma segura (comprueba usos antes) |

**Renombrar:** coloca el cursor en el símbolo, `Shift+F6`, escribe el nuevo nombre y confirma; PyCharm actualiza todos los usos en el proyecto.

## 6. Ejecutar y depurar

### Ejecución

| Atajo | Acción |
|-------|--------|
| **`Shift+F10`** | Ejecutar la configuración actual (Run) |
| **`Shift+F9`** | Ejecutar en modo depuración (Debug) |
| **`Ctrl+Shift+F10`** | Ejecutar el archivo o script bajo el cursor (crea config si no existe) |
| **`Ctrl+F5`** | Ejecutar sin depurar (Run) |

**Configuraciones de ejecución:** **Run → Edit Configurations**. Ahí se define el script a ejecutar, parámetros, variables de entorno, directorio de trabajo e interpretador de Python.

### Depuración (breakpoints y pasos)

| Atajo | Acción |
|-------|--------|
| **`Ctrl+F8`** | Poner/quitar breakpoint en la línea actual |
| **`F8`** | Step Over (ejecutar línea sin entrar en funciones) |
| **`F7`** | Step Into (entrar en la función llamada) |
| **`Shift+F8`** | Step Out (salir del frame actual) |
| **`F9`** | Resume (continuar hasta el siguiente breakpoint o fin) |
| **`Alt+F9`** | Run to Cursor (ejecutar hasta la línea donde está el cursor) |

En la ventana **Debug** se ven: variables (Variables), watchers, consola (Evaluate Expression con `Alt+F8` para evaluar expresiones).

## 7. Entornos virtuales e interpretador de Python

### Configurar el interpretador del proyecto

1. **File → Settings** (o **Ctrl+Alt+S**) → **Project: nombre → Python Interpreter**.
2. En **Python Interpreter** elegir:
   - Un venv existente (ruta a `python` dentro del venv).
   - **Add Interpreter → Add Local** y crear un nuevo entorno virtual (Virtualenv, Conda, etc.).

### Crear un venv desde PyCharm

- **Add Interpreter → Add Local → Virtualenv Environment**.
- **New**: ruta donde crear el venv (p. ej. `venv` o `.venv` en la raíz del proyecto).
- Base interpreter: Python del sistema o otra instalación.
- Opción **Make available to all projects** si quieres reutilizarlo.

### Instalar paquetes

- En la ventana **Python Interpreter**, botón **+** para instalar un paquete (PyPI).
- O en **Terminal** integrado (`Alt+F12`): `pip install nombre_paquete` (usa el interpretador del proyecto si el terminal está en el proyecto).

**Consejo:** Si el proyecto tiene `requirements.txt`, clic derecho sobre el archivo → **Install requirements**.

## 8. Git integrado

### Acciones desde la interfaz

| Acción | Dónde / Cómo |
|--------|----------------|
| Ver cambios | **Git → Commit** (o **Ctrl+K**) muestra archivos modificados y diff |
| Commit | Seleccionar archivos, escribir mensaje, **Commit** (o **Ctrl+Enter**) |
| Push | **Git → Push** (o **Ctrl+Shift+K**) |
| Pull | **Git → Pull** (o **Ctrl+T**) |

| Ramas | Clic en el nombre de rama (esquina inferior derecha) → **New Branch**, **Checkout**, etc. |
| Diff antes de commit | Clic derecho en archivo → **Compare with...** o en ventana Commit ver diff |

### Atajos útiles

| Atajo | Acción |
|-------|--------|
| **`Ctrl+K`** | Abrir ventana Commit |
| **`Ctrl+Shift+K`** | Push |
| **`Ctrl+T`** | Pull |
| **`Alt+9`** | Abrir ventana Git (cambios, log, etc.) |

**Commit:** se puede hacer **Commit and Push** en un solo paso desde la ventana de commit.

## 9. Jupyter Notebooks en PyCharm

- PyCharm permite abrir y editar archivos **`.ipynb`** (Jupyter).
- **Ejecutar celda:** botón ▶ junto a la celda, o **Shift+Enter** (ejecuta y pasa a la siguiente).
- **Ejecutar todas:** **Run → Run All** o desde el menú del notebook.
- **Interpretador:** el kernel usa el interpretador de Python configurado en el proyecto (o se puede elegir kernel en la esquina superior derecha del notebook).
- **Atajos en celdas:**
  - **`Ctrl+Enter`**: ejecutar celda actual (sin avanzar).
  - **`Alt+Enter`**: ejecutar celda e insertar nueva debajo.
  - **`Esc`** luego **A/B**: insertar celda arriba/abajo (en modo comando).
  - **`Esc`** luego **D D**: eliminar celda (en modo comando).
- **Depurar:** se puede poner breakpoint en una celda y depurar el notebook (Debug).

**Requisito:** tener instalado `jupyter` (y opcionalmente `ipykernel`) en el entorno del proyecto.

|----------------|-------------|
| **`Alt+F12`** | Abrir/cerrar la ventana Terminal |
| **Varias pestañas** | Botón **+** en la barra del terminal para nueva pestaña |

## 11. Estructura de proyecto y marcado de directorios

- **Raíz del proyecto:** la carpeta que abriste como proyecto. PyCharm indexa todo para búsquedas y refactors.
- **Marcar como Sources Root:** clic derecho en una carpeta → **Mark Directory as → Sources Root**. Esa carpeta se considera código fuente (imports desde la raíz del proyecto).
- **Excluir carpetas:** clic derecho → **Mark Directory as → Excluded**. PyCharm no indexa (útil para `venv`, `__pycache__`, `.git`, datos grandes). También en **Settings → Project Structure**.
- **Templates / Resources:** para proyectos web, se pueden marcar carpetas como Templates o Resource Roots.

**Consejo:** En proyectos Python, marcar la carpeta que contiene los paquetes como **Sources Root** para que los imports absolutos funcionen sin trucos.

## 12. Solución rápida de problemas frecuentes

| Problema | Qué hacer |
|----------|-----------|
| **No reconoce imports / subrayado rojo** | Comprobar **Python Interpreter** (Settings → Project → Python Interpreter). Debe apuntar al venv o entorno donde están instalados los paquetes. |
| **El terminal no usa el venv** | En **Settings → Tools → Terminal** comprobar que no se sobrescribe el activador. O cerrar y reabrir el terminal. |
| **Ejecutar no hace nada o error de config** | **Run → Edit Configurations** y definir un script (archivo .py) y el interpretador correcto. **Ctrl+Shift+F10** sobre el archivo crea una config automática. |
| **PyCharm va lento** | Excluir carpetas pesadas (datos, venv si no hace falta indexar). **File → Invalidate Caches** y reiniciar. Reducir inspecciones en Settings → Inspections. |
| **Atajos no funcionan** | Pueden estar en conflicto con el SO o con otro programa. **Settings → Keymap**, buscar la acción y asignar otro atajo. |
| **Jupyter no ejecuta** | Instalar `jupyter` e `ipykernel` en el interpretador del proyecto. En el notebook, elegir el kernel (esquina superior derecha) que coincida con ese interpretador. |

## 14. Configuración recomendada

- **Editor → General**: marcar **Save files automatically** si quieres guardado automático.
- **Editor → Code Style → Python**: ajustar indentación (normalmente 4 espacios), longitud de línea (p. ej. 100 o 88 para Black).
- **Tools → Python Integrated Tools**: definir **Default test runner** (pytest, unittest) para ejecutar tests con el botón verde junto a las pruebas.
- **Editor → Inspections**: revisar inspecciones de Python (PEP 8, tipo hints, etc.) y desactivar las que no interesen.
- **Plugins**: en **Settings → Plugins** se pueden instalar extensiones (Markdown, CSV, base de datos, etc.).

**Importar/exportar configuración:** **File → Manage IDE Settings → Export/Import** (útil para llevar la misma configuración a otro equipo).
